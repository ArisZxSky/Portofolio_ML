```python
#Type your code here
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler, LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix, classification_report
from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.naive_bayes import GaussianNB
import time
from sklearn.model_selection import GridSearchCV,  RandomizedSearchCV
from skopt import BayesSearchCV
```

# **2. Memuat Dataset dari Hasil Clustering**
Memuat dataset hasil clustering dari file CSV ke dalam variabel DataFrame.

```python
# Gunakan dataset hasil clustering yang memiliki fitur Target
# Silakan gunakan dataset data_clustering jika tidak menerapkan Interpretasi Hasil Clustering [Advanced]
# Silakan gunakan dataset data_clustering_inverse jika menerapkan Interpretasi Hasil Clustering [Advanced]
# Lengkapi kode berikut
url = 'https://raw.githubusercontent.com/ArisZxSky/datasets_dicoding/main/data_clustering_inverse.csv'
df = pd.read_csv(url)
```

```python
# Tampilkan 5 baris pertama dengan function head.
df.head()
```

# **3. Data Splitting**
Tahap Data Splitting bertujuan untuk memisahkan dataset menjadi dua bagian: data latih (training set) dan data uji (test set).

```python
# Encoder kolom kategori
fitur_kategori = df.select_dtypes(include=['object']).columns.tolist()
encoders = {}
df = pd.DataFrame(df)
for col in fitur_kategori:
    encoder = LabelEncoder()
    df[col] = encoder.fit_transform(df[col])
    encoders[col] = encoder

# Pisahkan fitur x dan y
x = df.drop(columns=['Target'], axis=1)
y = df['Target']

# Menggunakan train_test_split() untuk melakukan pembagian dataset.
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

# Menampilkan bentuk set pelatihan dan set uji untuk memastikan split
print(f'Training set shape: x_train ={x_train.shape}, y_train ={y_train.shape}')
print(f'test set shape: x_test={x_test.shape}, y_test = {y_test.shape}')
```

# **4. Membangun Model Klasifikasi**
Setelah memilih algoritma klasifikasi yang sesuai, langkah selanjutnya adalah melatih model menggunakan data latih.

Berikut adalah rekomendasi tahapannya.
1. Menggunakan algoritma klasifikasi yaitu Decision Tree.
2. Latih model menggunakan data yang sudah dipisah.

```python
# Buatlah model klasifikasi menggunakan Decision Tree
dt_model = DecisionTreeClassifier()
dt_model.fit(x_train, y_train)
```

```python
# Menyimpan Model
import joblib
joblib.dump(dt_model, 'decision_tree_model.h5')
```

# **5. Memenuhi Kriteria Skilled dan Advanced dalam Membangun Model Klasifikasi**



**Biarkan kosong jika tidak menerapkan kriteria skilled atau advanced**

```python
# Melatih model menggunakan algoritma klasifikasi scikit-learn selain Decision Tree.

# K-Nearst Neighbors
knn_model = KNeighborsClassifier()
knn_model.fit(x_train, y_train)

# Random Forest
rf_model = RandomForestClassifier()
rf_model.fit(x_train, y_train)

# Support Vector Machine (SVM)
svm_model = SVC()
svm_model.fit(x_train, y_train)

# Naive Bayes
nb_model = GaussianNB()
nb_model.fit(x_train, y_train)
```

```python
# Menampilkan hasil evaluasi akurasi, presisi, recall, dan F1-Score pada seluruh algoritma yang sudah dibuat.
def evaluate_model(model, x_test, y_test):
    y_pred = model.predict(x_test)
    cm = confusion_matrix(y_test, y_pred)
    results = {
        'Confusion Matrix' : cm,
        'Accuracy': accuracy_score(y_test, y_pred),
        'Precision' : precision_score(y_test, y_pred, average='macro'),
        'Recall': recall_score(y_test, y_pred, average='macro'),
        'F1-Score': f1_score(y_test, y_pred, average='macro'),
        'Classification Report': classification_report(y_test, y_pred)
    }
    return results

results = {
        'K-Nearest Neighbors (KNN)': evaluate_model(knn_model, x_test, y_test),
        'Decision Tree (DT)': evaluate_model(dt_model, x_test, y_test),
        'Random Forest (RF)': evaluate_model(rf_model, x_test, y_test),
        'Support Vector Machine (SVM)': evaluate_model(svm_model, x_test, y_test),
        'Naive Bayes (NB)': evaluate_model(nb_model, x_test, y_test)
}

# Buat DdataFrame untuk meringkas hasil
summary_df = pd.DataFrame(columns=['Model', 'Accuracy', 'Precision', 'Recall', 'F1-Score' ])

# Isi DataFrame dengan hasil
rows = []
for model_name, metrics in results.items():
    rows.append({
        'Model': model_name,
        'Accuracy': metrics['Accuracy'],
        'Precision': metrics['Precision'],
        'Recall': metrics['Recall'],
        'F1-Score': metrics['F1-Score']
     })


# Konversi daftar kamus ke DataFrame
summary_df = pd.DataFrame(rows)

# Tampilkan DataFrame
print(summary_df)
```

```python
# Menyimpan Model Selain Decision Tree
# Model ini bisa lebih dari satu
import joblib
joblib.dump(rf_model, 'explore_rf_classification.h5')
joblib.dump(knn_model, 'explore_knn_classification.h5')
joblib.dump(svm_model,'explore_svm_classification.h5')
joblib.dump(nb_model, 'explore_nb_classification.h5')
```

Hyperparameter Tuning Model

Pilih salah satu algoritma yang ingin Anda tuning

```python
# Lakukan Hyperparameter Tuning dan Latih ulang.
# Lakukan dalam satu cell ini saja.
start_time = time.time()  # Mencatat waktu mulai hyper parameter tuning

# Grid Search
# Mendefinisikan parameter
param_grid = {
    'max_depth' : [10, 20, 30],
    'min_samples_split' : [2, 5, 10],
    'min_samples_leaf' : [1, 2, 4],
    'criterion': ['gini', 'entropy']
}

grid_search = GridSearchCV(estimator=dt_model, param_grid=param_grid,
                           cv=3, n_jobs=-1, verbose=0)
grid_search.fit(x_train, y_train)


# Random Search
# Mendefinisikan parameter
param_dist = {
    'max_depth' : np.arange(5, 31, 5),
    'min_samples_split' : np.arange(2, 11, 2),
    'min_samples_leaf' : np.arange(1, 5, 1),
    'criterion': ['gini', 'entropy']
}

random_search = RandomizedSearchCV(estimator=dt_model, param_distributions=param_dist,
                                   n_iter=20, cv=3, n_jobs=-1, verbose=0, random_state=42)
random_search.fit(x_train, y_train)


# Bayesian Optimization
# Mendefinisikan parameter
bayes_search = BayesSearchCV(
    estimator=dt_model,
    search_spaces={
        'max_depth': (5, 30),
        'min_samples_split': (2, 10),
        'min_samples_leaf': (1, 5),
        'criterion': ['gini', 'entropy']
    },
    n_iter=20, cv=3, n_jobs=-1, random_state=42, verbose=0
)
bayes_search.fit(x_train, y_train)


```

```python
# Menampilkan hasil evaluasi akurasi, presisi, recall, dan F1-Score pada algoritma yang sudah dituning.

# Grid Search
best_dt_grid = grid_search.best_estimator_
# Accuracy
train_acc_grid = accuracy_score(y_train, best_dt_grid.predict(x_train))
test_acc_grid = accuracy_score(y_test, best_dt_grid.predict(x_test))
# Precision
train_prec_grid = precision_score(y_train, best_dt_grid.predict(x_train), average='macro')
test_prec_grid = precision_score(y_test, best_dt_grid.predict(x_test), average='macro')
# Recal
train_recall_grid = recall_score(y_train, best_dt_grid.predict(x_train), average='macro')
test_recall_grid = recall_score(y_test, best_dt_grid.predict(x_test), average='macro')
# F1-Score
train_f1_grid = f1_score(y_train, best_dt_grid.predict(x_train), average='macro')
test_f1_grid = f1_score(y_test, best_dt_grid.predict(x_test), average='macro')

print("Best Params:", grid_search.best_params_)
print(f"Train Acc: {train_acc_grid:.2f} | Test Acc: {test_acc_grid:.2f}")
print(f'Train Precision: {train_prec_grid:.2f} | Test Precision: {test_prec_grid:.2f}')
print(f'Train Recall: {train_recall_grid:.2f} | Test Recall: {test_recall_grid:.2f}')
print(f'Train F1-Score: {train_f1_grid:.2f} | Test F1-Score: {test_f1_grid:.2f}')

# Random Search
best_dt_random = random_search.best_estimator_
# Accuracy
train_acc_random = accuracy_score(y_train, best_dt_random.predict(x_train))
test_acc_random = accuracy_score(y_test, best_dt_random.predict(x_test))
# Precision
train_prec_random = precision_score(y_train, best_dt_random.predict(x_train), average='macro')
test_prec_random = precision_score(y_test, best_dt_random.predict(x_test), average='macro')
# Recal
train_recall_random = recall_score(y_train, best_dt_random.predict(x_train), average='macro')
test_recall_random = recall_score(y_test, best_dt_random.predict(x_test), average='macro')
# F1-Score
train_f1_random = f1_score(y_train, best_dt_random.predict(x_train), average='macro')
test_f1_random = f1_score(y_test, best_dt_random.predict(x_test), average='macro')

print("Best Params:", random_search.best_params_)
print(f"Train Acc: {train_acc_random:.2f} | Test Acc: {test_acc_random:.2f}")
print(f'Train Precision: {train_prec_random:.2f} | Test Precision: {test_prec_random:.2f}')
print(f'Train Recall: {train_recall_random:.2f} | Test Recall: {test_recall_random:.2f}')
print(f'Train F1-Score: {train_f1_random:.2f} | Test F1-Score: {test_f1_random:.2f}')

# Bayesian Optimization
best_dt_bayes = bayes_search.best_estimator_
# Accuracy
train_acc_bayes = accuracy_score(y_train, best_dt_bayes.predict(x_train))
test_acc_bayes = accuracy_score(y_test, best_dt_bayes.predict(x_test))
# Precision
train_prec_bayes = precision_score(y_train, best_dt_bayes.predict(x_train), average='macro')
test_prec_bayes = precision_score(y_test, best_dt_bayes.predict(x_test), average='macro')
# Recal
train_recall_bayes = recall_score(y_train, best_dt_bayes.predict(x_train), average='macro')
test_recall_bayes = recall_score(y_test, best_dt_bayes.predict(x_test), average='macro')
# F1-Score
train_f1_bayes = f1_score(y_train, best_dt_bayes.predict(x_train), average='macro')
test_f1_bayes = f1_score(y_test, best_dt_bayes.predict(x_test), average='macro')

print("Best Params:", bayes_search.best_params_)
print(f"Train Acc: {train_acc_bayes:.2f} | Test Acc: {test_acc_bayes:.2f}")
print(f'Train Precision: {train_prec_bayes:.2f} | Test Precision: {test_prec_bayes:.2f}')
print(f'Train Recall: {train_recall_bayes:.2f} | Test Recall: {test_recall_bayes:.2f}')
print(f'Train F1-Score: {train_f1_bayes:.2f} | Test F1-Score: {test_f1_bayes:.2f}')
```
